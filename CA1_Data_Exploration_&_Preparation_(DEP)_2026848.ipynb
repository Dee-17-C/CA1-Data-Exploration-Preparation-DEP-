## 1. Problem Domain & Motivation
The global credit card industry faces significant financial risks due to customer defaults. Accurately predicting the probability of default is a critical challenge in the **Economy and Finance** domain. The motivation for this project is to explore historical credit card client data to identify demographic and behavioral risk factors (covariations and variations) associated with defaults. Furthermore, handling high-dimensional financial tracking data is computationally expensive. By exploring this dataset, the goal is to prepare the data for future predictive modeling by cleaning anomalies, applying feature engineering, and reducing dimensionality without losing critical financial variance.

**Dataset Chosen:** "Default of Credit Card Clients" from the UCI Machine Learning Repository.

https://archive.ics.uci.edu/dataset/350/default+of+credit+card+clients


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.decomposition import PCA

import warnings
warnings.filterwarnings('ignore')

# 1. LOAD DATASET
# The UCI dataset uses the first row for metadata, so header=1 is required
file_path = "default of credit card clients.xls"
df = pd.read_excel(file_path, header=1)

print(f"Original Dataset Shape: {df.shape}")
display(df.head())
```

## 2. Dataset Characterization & Data Preparation
The dataset loaded successfully with 30,000 rows and 25 columns.
**Preparation Steps:**
1. **Redundant Columns:** Drop the `ID` column as it holds no analytical value.
2. **Out-of-Range Data:** Undocumented categories in `EDUCATION` (0, 5, 6) and `MARRIAGE` (0) act as corrupted data. We map them to their respective "Others" categories (4 and 3).
3. **Outliers:** `LIMIT_BAL` (credit limit) contains extreme outliers. We filter out values above the 99th percentile to normalize the distribution.

```python
# Drop ID column
if 'ID' in df.columns:
    df = df.drop(columns=['ID'])

# Check for missing NaN values
print(f"Missing NaN values in dataset: {df.isnull().sum().sum()}")

# Handle Out-of-range Categorical Data
df['EDUCATION'] = df['EDUCATION'].replace([0, 5, 6], 4)
df['MARRIAGE'] = df['MARRIAGE'].replace(0, 3)

# Handle Outliers in LIMIT_BAL (99th Percentile)
limit_99th = df['LIMIT_BAL'].quantile(0.99)
df_clean = df[df['LIMIT_BAL'] <= limit_99th]

print(f"Cleaned Dataset Shape: {df_clean.shape}")
```

## 3. Exploratory Data Analysis (EDA) & Unusual Patterns
To justify our feature engineering and PCA steps, we formulate four analytical questions:
* **Question 1 (Variation):** How does the credit limit (`LIMIT_BAL`) vary between clients who default and those who do not?
* **Question 2 (Covariation):** Is there an unusual pattern or high redundancy across the 6-month historical billing data (`BILL_AMT1` to `BILL_AMT6`)?
* **Question 3 (Variation):** How do demographic attributes like Education and Age vary between defaulters and non-defaulters?
* **Question 4 (Covariation):** Do the historical payment amounts (`PAY_AMT1` to `PAY_AMT6`) exhibit correlations that further justify their inclusion in the dimensionality reduction step?

```python
# Set visual style
sns.set_theme(style="whitegrid")

# 1. Variation: Credit Limit Distribution by Default Status
plt.figure(figsize=(10, 5))
sns.histplot(data=df_clean, x='LIMIT_BAL', hue='default payment next month',
             kde=True, bins=30, palette="Set2")
plt.title('Variation in Credit Limit (LIMIT_BAL) by Default Status', fontsize=14)
plt.xlabel('Credit Limit')
plt.ylabel('Frequency')
plt.show()

# 2. Covariation & Unusual Patterns: Heatmap of Billing Amounts
bill_cols = ['BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6']
plt.figure(figsize=(8, 6))
correlation_matrix = df_clean[bill_cols].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Covariation between 6-Month Bill Amounts', fontsize=14)
plt.show()

print("High covariation observed (>0.80) across all billing months, justifying dimensionality reduction.")

# 3. Variation: Default Rates by Education Level (Demographic Factor)
plt.figure(figsize=(8, 5))
ax = sns.countplot(data=df_clean, x='EDUCATION', hue='default payment next month', palette='Set2')
plt.title('Variation in Default Status by Education Level', fontsize=14)
plt.xlabel('Education Level (1=Grad, 2=Uni, 3=High School, 4=Others)')
plt.ylabel('Count')
plt.legend(title='Default Next Month')
plt.show()

# 4. Variation: Age Distribution of Defaulters vs Non-Defaulters
plt.figure(figsize=(8, 5))
sns.boxplot(data=df_clean, x='default payment next month', y='AGE', palette='Set2')
plt.title('Age Distribution by Default Status', fontsize=14)
plt.xlabel('Default Next Month (0 = No, 1 = Yes)')
plt.ylabel('Age')
plt.show()

# 5. Covariation: Heatmap of Previous Payment Amounts (PAY_AMT)
# Justifies including PAY_AMT in the PCA alongside BILL_AMT
pay_cols = ['PAY_AMT1', 'PAY_AMT2', 'PAY_AMT3', 'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']
plt.figure(figsize=(8, 6))
correlation_matrix_pay = df_clean[pay_cols].corr()
sns.heatmap(correlation_matrix_pay, annot=True, cmap='viridis', fmt=".2f", linewidths=0.5)
plt.title('Covariation between 6-Month Payment Amounts', fontsize=14)
plt.show()
```

**EDA Findings & Interpretation:**
* **Demographics (Age & Education):** The boxplot reveals that the median age for both defaulters and non-defaulters is identical (34 years old), suggesting age is not a strong standalone predictor of default. The education count plot shows that University (Level 2) and Graduate School (Level 1) clients make up the vast majority of the dataset.
* **Financial Covariation:** The billing heatmap confirmed extreme covariation (>0.80) across all 6 billing months. The payment heatmap (`PAY_AMT`) showed much weaker covariation (mostly <0.25). Regardless, PCA is justified to compress the highly redundant billing data alongside the payment data to streamline the continuous feature space.
