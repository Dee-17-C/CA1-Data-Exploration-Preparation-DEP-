## 1. Problem Domain & Motivation
The global credit card industry faces significant financial risks due to customer defaults. Accurately predicting the probability of default is a critical challenge in the **Economy and Finance** domain. The motivation for this project is to explore historical credit card client data to identify demographic and behavioral risk factors (covariations and variations) associated with defaults. Furthermore, handling high-dimensional financial tracking data is computationally expensive. By exploring this dataset, the goal is to prepare the data for future predictive modeling by cleaning anomalies, applying feature engineering, and reducing dimensionality without losing critical financial variance.

**Dataset Chosen:** "Default of Credit Card Clients" from the UCI Machine Learning Repository.

https://archive.ics.uci.edu/dataset/350/default+of+credit+card+clients


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.decomposition import PCA

import warnings
warnings.filterwarnings('ignore')

# 1. LOAD DATASET
# The UCI dataset uses the first row for metadata, so header=1 is required
file_path = "default of credit card clients.xls"
df = pd.read_excel(file_path, header=1)

print(f"Original Dataset Shape: {df.shape}")
display(df.head())
```

## 2. Dataset Characterization & Data Preparation
The dataset loaded successfully with 30,000 rows and 25 columns.
**Preparation Steps:**
1. **Redundant Columns:** Drop the `ID` column as it holds no analytical value.
2. **Out-of-Range Data:** Undocumented categories in `EDUCATION` (0, 5, 6) and `MARRIAGE` (0) act as corrupted data. We map them to their respective "Others" categories (4 and 3).
3. **Outliers:** `LIMIT_BAL` (credit limit) contains extreme outliers. We filter out values above the 99th percentile to normalize the distribution.

```python
# Drop ID column
if 'ID' in df.columns:
    df = df.drop(columns=['ID'])

# Check for missing NaN values
print(f"Missing NaN values in dataset: {df.isnull().sum().sum()}")

# Handle Out-of-range Categorical Data
df['EDUCATION'] = df['EDUCATION'].replace([0, 5, 6], 4)
df['MARRIAGE'] = df['MARRIAGE'].replace(0, 3)

# Handle Outliers in LIMIT_BAL (99th Percentile)
limit_99th = df['LIMIT_BAL'].quantile(0.99)
df_clean = df[df['LIMIT_BAL'] <= limit_99th]

print(f"Cleaned Dataset Shape: {df_clean.shape}")
```
